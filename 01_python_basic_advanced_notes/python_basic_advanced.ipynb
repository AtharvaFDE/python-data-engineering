# Python Type Casting

Type Casting is the process of converting one data type into another data type in Python.

## 1. Types of Type Casting

### A. Implicit Type Casting
Python automatically converts one data type into another without requiring manual conversion.

**Example:**
```python
a = 10
b = 5.5

result = a + b

print(result)        # 15.5
print(type(result))  # <class 'float'>
```

### B. Explicit Type Casting
Explicit Type Casting means manually converting one data type into another using built-in functions.

**Example:**
```python
a = "100"

b = int(a)

print(b)        # 100
print(type(b))  # <class 'int'>
```

## 2. Type Casting Functions

| Function | Description | Example |
|---|---|---|
| `int()` | Converts into Integer | `int("10")` |
| `float()` | Converts into Float | `float("10.5")` |
| `str()` | Converts into String | `str(100)` |
| `bool()` | Converts into Boolean | `bool(1)` |
| `list()` | Converts into List | `list((1, 2, 3))` |
| `tuple()` | Converts into Tuple | `tuple([1, 2, 3])` |
| `set()` | Converts into Set | `set([1, 2, 2])` |
| `dict()` | Converts into Dictionary from valid key-value pairs | `dict([("a", 1)])` |

## 3. Example of Type Casting

```python
a = "25"

print(int(a))    # 25
print(float(a))  # 25.0
print(bool(a))   # True
```

## Important Points

- Implicit casting is performed automatically by Python.
- Explicit casting is performed manually by the programmer.
- Invalid conversions can raise a `ValueError`.
- Type casting helps perform operations between different data types.

- Logical operators are commonly used with conditional statements.


# Python Conditions and If-ELSE Statements

Conditional statements are used to execute different blocks of code based on whether a condition is True or False.

## 6. Python Conditions and If-ELSE Statements

### 1. IF Statements

Executes a block of code if the given condition is True.

```python
age = 24

if age >= 18:
    print("Eligible to vote")
```

### 2. IF-ELSE Statements

Executes the `if` block when the condition is True; otherwise, executes the `else` block.

```python
age = 16

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

### 3. IF-ELIF-ELSE Statements

Used to check multiple conditions.

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

### 4. Nested IF Statements

An if statement inside another if statement is called a Nested IF statement.

```python
age = 24
has_id = True

if age >= 18:
    if has_id:
        print("Entry Allowed")
    else:
        print("ID Required")
else:
    print("Entry Not Allowed")
```

## Important Points

- `if` checks a condition.
- `else` executes when the if condition is False.
- `elif` checks additional conditions.
- Nested if means an if statement inside another if statement.
- Python uses indentation to define code blocks.


# Python For Loop

A `for` loop is used for iterating over a sequence (such as a list, tuple, dictionary, set, or string).

The `for` loop executes a set of statements once for each item in a sequence.

## 7. For Loop

### 1. Looping Through List

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

**Output:**
```text
apple
banana
cherry
```

### 2. Looping Through Strings

```python
name = "Python"

for letter in name:
    print(letter)
```

**Output:**
```text
P
y
t
h
o
n
```

### 3. The Break Statement

The `break` statement is used to terminate the loop when a condition is met.

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

**Output:**
```text
1
2
```

### 4. The Continue Statement

The `continue` statement skips the current iteration and moves to the next iteration.

**Print all letters except 'l':**

```python
name = "Hello"

for letter in name:
    if letter == 'l':
        continue
    print(letter)
```

**Output:**
```text
H
e
o
```

### 5. The range() Function

The `range()` function generates a sequence of numbers.

**Syntax:**
```python
range(start, stop, step)
```

**Example:**

```python
for i in range(1, 6):
    print(i)
```

**Output:**
```text
1
2
3
4
5
```

## Important Points

- `for` loop iterates over a sequence.
- `break` terminates the loop.
- `continue` skips the current iteration.
- `range()` generates a sequence of numbers.
- Python uses indentation to define the loop body.


````python
# Looping through List 

fruits = ["apple","banana","cherry"]

for i in fruits:
    print('eat',i)
````


# Python Nested Loops

A Nested Loop is a loop inside another loop. The inner loop executes completely for every single iteration of the outer loop.

## 8. Nested Loops

### 1. Nested For Loop

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

**Output:**
```text
1 1
1 2
1 3
2 1
2 2
2 3
3 1
3 2
3 3
```

### 2. Nested Loop with Pattern

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

**Output:**
```text
*
* *
* * *
* * * *
* * * * *
```

### 3. Nested Loop with Multiplication Table

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()
```

**Output:**
```text
1 2 3
2 4 6
3 6 9
```

## Important Points

- A loop inside another loop is called a Nested Loop.
- The inner loop executes completely for every iteration of the outer loop.
- Nested loops are commonly used for patterns, matrices, and tables.
- We can use nested `for` loops and nested `while` loops.


````python
# Nested Loops 
````

````python
x = [1,3,45,6]
y = [7,9,5]

for i in x:
    for j in y:
        print(i,j)
````


# Python While Loop

A `while` loop is used to execute a set of statements repeatedly as long as a given condition is True.

Python has two primitive loop commands:
- `while` loops
- `for` loops

## 8. While Loop

### 1. Basic While Loop

Print numbers as long as `i` is less than 6.

```python
i = 1

while i < 6:
    print(i)
    i += 1
```

**Output:**
```text
1
2
3
4
5
```

### 2. While Loop with Break Statement

The `break` statement is used to stop the loop when a condition is met.

```python
i = 1

while i < 6:
    if i == 3:
        break
    print(i)
    i += 1
```

**Output:**
```text
1
2
```

### 3. While Loop with Continue Statement

The `continue` statement skips the current iteration and moves to the next iteration.

```python
i = 0

while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

**Output:**
```text
1
2
4
5
```

## Important Points

- A `while` loop executes as long as the condition is True.
- `break` terminates the loop.
- `continue` skips the current iteration.
- Always update the loop variable to avoid an infinite loop.
- Python uses indentation to define the loop body.


````python
i = 1

while i <= 6:
    print(i)
    i+=1
````

````python
i = 1

while i<=6:
    if i == 3:
        break;
    print(i)
    i+=1
````
