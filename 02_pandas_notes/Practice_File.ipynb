# Merging , Joining and Concatenating

```python
s1 = pd.Series(data=['A','B','A','B','C'])
s2 = pd.Series(data=[11,22,33,44,55])
data2 = pd.DataFrame({
    'Name':s1,
    'Lemonades':s2
})
```

```python
data2
```

```python
s1 = pd.Series(data=['A','B','A','B','C'])
s2 = pd.Series(data=[10,20,30,40,50])
data1 = pd.DataFrame({
    'Name':s1,
    'Lemon':s2
})

data1
```

```python
pd.concat([data1,data2])
```

```python
data1.append(data2)  # Append is being removed From a Python 2 Version 
```

```python
pd.merge(data1, data2, on ="Name")
```

```python
data1.join(data2.set_index('Name'), on='Name')
```
